# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# TOOL 3: Check Palindrome

def is_palindrome(text: str) -> bool:
    """Check if a string is a palindrome."""
    try:
        return text == text[::-1]
    except Exception:
        return False
````

## Logging Setup

````python
import logging

logging.basicConfig(
    filename = "agent.log",
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    datefmt="%H:%M:%S",
)
logger = logging.getLogger("agent")
````

## Improving Routing

````python
import re

def detect_intent(query):

    query = query.lower()

    if re.search(r"\d+\s*[\+\-\*/]\s*\d+", query):
        return "calculation"

    if "keywords" in query or "extract" in query:
        return "keywords"

    if "palindrome" in query or "check palindrome" in query:
        return "palindrome"

    return "general"
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)

def agent(query):
    try:
        intent = detect_intent(query)

        # TOOL1 CALC
        if intent == "calculation":
            expression = query.lower()
            for word in ["calculate","compute","solve"]:
                expression = expression.replace(word, "")
            expression = expression.strip()
            logger.info(f"Calculation query: {expression}")
            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        # TOOL2 Extract Keyword
        elif intent == "keywords":
            text = query.lower()
            text = text.replace("extract keywords from", "")
            text = text.replace("keywords from", "")
            text = text.strip()
            logger.info(f"Keyword query: {text}")
            return {
                "type": "keywords",
                "result": extract_keywords(text)
            }

        # TOOL3 Palindrome
        elif intent == "palindrome":
            text = query.lower()
            text = text.replace("check palindrome", "")
            text = text.strip()
            logger.info(f"Palindrome query: {text}")
            return {
                "type": "palindrome",
                "result": is_palindrome(text)
            }

        # General
        else:
            logger.info(f"General query: {query}")
            return {
                "type": "general",
                "result": f"This was a General query: {query}"
            }

    except Exception as e:
        logger.error(f"Error in agent: {e}")
        return {
            "type": "error",
            "result": str(e)
        }
````
